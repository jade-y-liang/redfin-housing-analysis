# 0. Word of Advice

When tackling a data science project, especially one involving a dataset like housing prices, a structured approach is crucial for success. Here’s a step-by-step guide that you can follow to clean your data, analyze it for insights, and build a predictive model:

1. Define the Problem

Before diving into the data, clarify what you’re trying to solve. In your case, it seems you want to predict housing prices based on details about each house. This is typically a regression problem.

2. Data Collection

It sounds like you already have the dataset, but this step typically involves gathering the necessary data from various sources which could be databases, CSV files, APIs, or other formats.

3. Data Cleaning

Given that you’ve removed rows with all NaN values, there are likely more cleaning tasks to perform:
	•	Handle Missing Data: For remaining missing values, decide whether to fill them (e.g., with the mean, median, or mode of the column), or remove rows/columns with too many missing values.
	•	Remove Duplicates: Check for and remove any duplicate entries to prevent skewed analysis.
	•	Correct Data Types: Ensure each column is of the correct data type (e.g., numerical, categorical).
	•	Remove Outliers: Identify and remove outliers that could skew your analysis and model.

4. Exploratory Data Analysis (EDA)

This is a crucial step to understand the patterns, anomalies, hypothesis testing, and relationships within your data.
	•	Univariate Analysis: Study the distribution of each variable, using plots like histograms, box plots, or density plots.
	•	Bivariate/Multivariate Analysis: Analyze relationships between variables using scatter plots, correlation matrices, and cross-tabulations.
	•	Feature Engineering: Create new features that might be useful for the model, based on your understanding of the data and domain knowledge.

5. Data Preprocessing

Prepare your data for modeling:
	•	Encode Categorical Variables: Convert categorical variables using one-hot encoding or label encoding.
	•	Feature Scaling: Standardize or normalize features, especially if using algorithms sensitive to the scale of input data (like SVM, k-nearest neighbors, or neural networks).
	•	Split the Data: Divide your data into training and testing sets (common splits are 70:30, 80:20).

6. Model Building

Select and build appropriate models. Since it’s a regression problem:
	•	Baseline Model: Start with simple models like linear regression to establish a baseline.
	•	Experiment with Different Models: Use various algorithms like decision trees, random forest, and gradient boosting machines (GBMs). If data and context allow, experiment with more sophisticated models like XGBoost or neural networks.
	•	Cross-Validation: Implement k-fold cross-validation to validate model performance reliably.

7. Model Evaluation

Evaluate model performance using appropriate metrics:
	•	Regression Metrics: Use metrics like Mean Absolute Error (MAE), Mean Squared Error (MSE), or R-squared to judge model accuracy.
	•	Compare Models: Assess different models against your baseline to find the best performer.

8. Model Tuning

Optimize the model:
	•	Hyperparameter Tuning: Use techniques like Grid Search or Random Search to find the optimal settings for model parameters.
	•	Feature Selection: Identify the most important features contributing to the outcome and adjust your model accordingly.

9. Visualization of Results

Visualize predictions and model performance clearly to communicate your findings effectively:
	•	Plot Predictions vs. Actuals: Create plots that compare predicted values against actual values.
	•	Feature Importance: Visualize the importance of different features in the model.

10. Deployment (Optional)

If the model performs well, you might deploy it to a production environment:
	•	Integration: Integrate the model into an existing application or workflow.
	•	Monitoring and Maintenance: Set up systems to monitor model performance over time and recalibrate as necessary with new data.

11. Documentation and Reporting

Finally, document your findings, methodologies, and any issues encountered during the project. Prepare a report or presentation to communicate the results to stakeholders.

Following these steps systematically will help ensure that your project is thorough, the insights are reliable, and the predictive models are robust.

With the additional context that each zip code only contains a few hundred rows, and some even less, the challenge of accurately predicting housing prices can intensify due to the limited data per zip code. Here are some strategies and considerations to effectively handle this situation:

1. Pooling Data Across Zip Codes

If feasible, consider pooling data from multiple zip codes to increase the dataset size for each model training session. This approach can help in two ways:
	•	Statistical Strength: More data can provide more statistical validity to the models, helping reduce overfitting and increasing the robustness of predictions.
	•	Generalization: By training on a diverse set of data covering multiple areas, the model may better learn to generalize across different types of houses and markets.

2. Hierarchical or Mixed Models

You might consider hierarchical modeling, where zip codes are random effects within a broader model. This allows:
	•	Shared Learning: The model can learn common patterns across all data while also adapting to the specifics of individual zip codes.
	•	Better Estimates for Sparse Data: For zip codes with very few data points, hierarchical models can make more reliable predictions by borrowing strength from the overall patterns learned across larger samples.

3. Specialized Models for Each Zip Code

If there’s enough variability between zip codes, it might make sense to train individual models for each zip code. However, this approach is limited by:
	•	Data Scarcity: Some zip codes with very few data points might not support a robust individual model.
	•	Overfitting Risk: Smaller datasets are inherently more prone to overfitting, especially if a complex model is used.

4. Feature Engineering

Enhance your dataset with:
	•	External Data: Integrate more features that could influence housing prices, such as economic indicators, demographics of the area, or real estate trends.
	•	Domain Expertise: Use insights from real estate experts to create derived features that are more informative and possibly compensate for the smaller dataset size.

5. Dimensionality Reduction

For areas with sparse data, reducing the number of features (either through feature selection or dimensionality reduction techniques like PCA) can help mitigate overfitting by simplifying the model.

6. Robust Validation Techniques

Implement robust validation methods to ensure the model performs well despite the limited data:
	•	Cross-validation: Use techniques like k-fold cross-validation within each zip code if possible, to maximize the use of available data.
	•	Bootstrapping: Especially in zip codes with very few samples, use bootstrapping to simulate a larger dataset for model training and validation.

7. Model Simplicity

Start with simpler, less complex models to establish a baseline:
	•	Linear Models: These can be quite effective and less likely to overfit when data is limited.
	•	Regularization: Use models with built-in regularization (like Ridge or Lasso regression) to prevent overfitting.

8. Continuous Monitoring and Updates

Once a model is developed, it’s crucial to monitor its performance continuously and update it as more data becomes available:
	•	Feedback Loop: Implement a system to gather feedback on the model’s predictions and use this data to refine and improve the model over time.

Summary

Working with such segmented and limited data requires careful consideration of the modeling approach, robust validation to ensure the model’s applicability, and possibly the integration of external data to enhance the dataset’s richness and reliability. Each strategy has its trade-offs and should be chosen based on the specific characteristics of your data and the business objectives.

```python
# loading necessary packages
import pandas as pd
import glob
```

```python
# load properties downloaded from Redfin
csv_files = glob.glob('nov24_la_datasets/' + "*.csv")

# initialize list to store dataframes
dfs = []

# loop through each csv file and append to dfs
for file in csv_files:
    df = pd.read_csv(file)
    dfs.append(df)

# merge all dataframes together
raw_nov_properties = pd.concat(dfs, ignore_index=True)
print(f"There is a total of {len(raw_nov_properties)} properties on sale.")
raw_nov_properties.head()
```

```python
# load properties downloaded from Redfin
csv_files = glob.glob('jan25_la_datasets/' + "*.csv")

# initialize list to store dataframes
dfs = []

# loop through each csv file and append to dfs
for file in csv_files:
    df = pd.read_csv(file)
    dfs.append(df)

# merge all dataframes together
raw_jan_properties = pd.concat(dfs, ignore_index=True)
print(f"There is a total of {len(raw_jan_properties)} properties on sale.")
raw_jan_properties.head()
```

# 3. Data Cleaning

```python
def clean_properties(raw_properties):
    """ cleans Redfin properties dataset by dropping unnecessary columns, removing rows
        with all missing values, removing duplicate rows, standardizing text data by 
        converting to lowercase, and converting numeric values from float to int

    Args:
        raw_properties (pandas DataFrame): the raw Redfin dataset

    Returns:
        properties: a pandas DataFrame of the cleaned data
    """

    properties = raw_properties.copy()

    # dropping unnecessary columns
    properties = properties.drop(columns=["SALE TYPE", "SOLD DATE", "STATE OR PROVINCE", "NEXT OPEN HOUSE START TIME",
                                          "STATUS", "NEXT OPEN HOUSE END TIME","SOURCE", "MLS#", "FAVORITE", "INTERESTED"])
    
    properties = properties.dropna(how='all') # removing rows with all missing values 
    properties = properties.drop_duplicates(ignore_index=True) # removing duplicate rows


    # renaming columns for convenience
    properties.columns = properties.columns.str.lower() # convert all columns to lowercase
    properties.columns = properties.columns.str.replace(" ", "_", regex=True) # replacing space with underscore
    properties = properties.rename(columns={"url_(see_https://www.redfin.com/buy-a-home/comparative-market-analysis_for_info_on_pricing)" : 'url',
                                            "zip_or_postal_code": "zip_code",
                                            "state_or_province": "state",
                                            "$/square_feet": "cost_per_sq_feet",
                                            "hoa/month": "monthly_hoa_cost"
                                            })

    # convert numeric columns from float to int
    numeric_cols = ['zip_code', 'price', 'beds', 'baths', 'square_feet', 'lot_size', 'year_built',
                    'days_on_market', 'cost_per_sq_feet', 'monthly_hoa_cost']
    for col in numeric_cols: properties[col] = properties[col].fillna(0).astype(int) # replaces na with 0

    # remove duplicate rows
    properties = properties.drop_duplicates()

    # remove row where zip code is 0 (only one such row in dataset)
    properties = properties[properties["zip_code"] != 0]

    return properties

# cleaning dataset
nov_properties, jan_properties = raw_nov_properties.copy(), raw_jan_properties.copy()
nov_properties, jan_properties = clean_properties(nov_properties), clean_properties(jan_properties)
nov_properties.head(10)
```

# 5. Exploratory Data Analysis

As we can see here, Downtown L.A. has the most amount of houses for sale in November 2024 on `Redfin.com`.

```python
import plotly.express as px
NOV_NUM_OF_PROPERTIES = len(nov_properties)

# count number of houses in each location
nov_location_count = nov_properties.groupby(["location"]).size().reset_index(name = "count").sort_values(by="count", ascending=True)

# evaluate percentage of houses in each location
nov_location_count["percentage"] = (nov_location_count["count"]/NOV_NUM_OF_PROPERTIES).round(5)
nov_location_count.head()

# generate bar chart to see percentage of houses from each neighborhood
fig = px.bar(nov_location_count, x = "count", y = "location", 
             hover_name = "location", hover_data = ["percentage", "count"],
             height=2000, color = "percentage", color_continuous_scale = "sunset",
             title = "Percentage of Houses from Each Location (Nov 2024)",
             labels = {"location": "Location",
                       "count": "Count"}
             )
fig.update_layout(bargap=0.25)
fig.show()
```
