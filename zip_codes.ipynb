```python
import pandas as pd
zip_codes = pd.read_csv("zip_code_database.csv")
zip_codes.head()
```

```python
ca_zip_codes = zip_codes[zip_codes["state"] == "CA"]
ca_zip_codes.head()
```

```python
la_zip_codes = ca_zip_codes[ca_zip_codes["primary_city"] == "Los Angeles"]
la_zip_codes
```

```python
# saving LA zip codes as a CSV file
la_zip_codes.to_csv("la_zip_codes.csv")
```

```python
la_county_zip_codes = ca_zip_codes[ca_zip_codes["county"] == "Los Angeles County"]
print(len(la_county_zip_codes))
la_county_zip_codes.head()
```
